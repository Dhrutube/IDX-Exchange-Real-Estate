```python
'''
some columns seem to have a column.1 name
'''

# checking if they are duplicates for listings
for col in listingsdf.columns:
    if col[-2:] == '.1':
        print(f'Is {col} a duplicate?\n{listingsdf[col[:-2]].equals(listingsdf[col])}')

print()

# checking if they are duplicates for sold
for col in solddf.columns:
    if col[-2:] == '.1':
        print(f'Is {col} a duplicate?\n{solddf[col[:-2]].equals(solddf[col])}')
```

```python
# seems they are only duplicates for listings, not sold
# checking distributions
for col in solddf.columns:
    if col[-2:] == '.1':
        print(f'Null values in {col[:-2]}:', solddf[col[:-2]].isna().sum())
        print(f'Null values in {col}:', solddf[col].isna().sum())
        print()
# orders of magnitude more missing values in column.1
# dropped duplicates in listingsdf in script.py
```

```python
# key columns as defined by the primer
# these columns should not be dropped even if there are a large number of missing values
keycols = set(('ListingKey', 
           'ListingContractDate', 
           'ListPrice',
           'ClosePrice',
           'PurchaseContractDate',
           'CloseDate',
           'LivingArea',
           'BedroomsTotal',
           'BathroomsTotalInteger',
           'Longitude',
           'Latitude',
           'UnparsedAddress',
           'ListAgentAOR'
           'DaysOnMarket'))
```

```python
# identifying columns with >90% missing values

# listings
for col in listingsdf.columns:
    if listingsdf[col].isna().sum() > 0.90:
        if col in keycols:
            print(f'Key column in Listings with >90% missing values: {col}')
            continue
        listingsdf.drop(columns=[col], inplace=True)

print()

# sold
for col in solddf.columns:
    if solddf[col].isna().sum() > 0.90:
        if col in keycols:
            print(f'Key column in Sold with >90% missing values: {col}')
            continue
        solddf.drop(columns=[col], inplace=True)   
```

```python
# CloseDate and ClosePrice have a large number of missing values
# The Sold dataframe has a large number of missing values in key columns. 
# This makes sense because not every listing would have been sold yet. 
# The data can be extracted from the listings dataframe for now.
```

```python
toDropL = set()
for col in listingsdf.columns:
    if col[-2:] == '.1':
        toDropL.add(col)
    if listingsdf[col].isna().sum() > 0.90:
        if col in keycols:
            print(f'Key column in Listings with >90% missing values: {col}')
            continue
        else:
            toDropL.add(col)
listingsdf.drop(columns=toDropL, inplace=True)
```

```python
toDropS = set()
for col in solddf.columns:
    if col[-2:] == '.1':
        toDropS.add(col)
    if solddf[col].isna().sum() > 0.90:
        if col in keycols:
            print(f'Key column in Listings with >90% missing values: {col}')
            continue
        else:
            toDropS.add(col)
solddf.drop(columns=toDropS, inplace=True)
```
